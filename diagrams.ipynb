```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('healthcare_dataset.csv')
df.head()
```

```python
x = df['Blood Type'].unique()
y = df['Blood Type'].value_counts()
```

```python
plt.bar(x,y, color='orange', label='No of Patients')
plt.xlabel('Blood Type')
plt.ylabel('Number Of Patients')
plt.title('Number of Patient with Blood Type')
plt.legend()
plt.ylim(1000, 10000)
plt.show()
```

```python
plt.barh(x, y, color='orange', label='No of Patients')
plt.xlabel('Number Of Patients')
plt.ylabel('Blood Type')
plt.title('Number of Patient with Blood Type')
plt.show()
```

```python
labels = df['Admission Type'].unique()
values = df['Admission Type'].value_counts()
```

```python
plt.pie(values, labels=labels, autopct='%1.1f%%', colors=['red', 'orange', 'yellow'])
plt.title('Admission Type Chart')
plt.show()
```

```python
ages = df['Age']
```

```python
plt.hist(ages, bins=10, edgecolor='red', color='orange')
plt.title('Ages Range')
plt.xlabel('Ages')
plt.ylabel('No of Patients')
plt.show()
```
