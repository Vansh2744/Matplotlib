```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('netflix_titles.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
types_count = df['type'].value_counts()
```

```python
plt.bar(types_count.index, types_count.values, color=['orange', 'orangered'], label='movies vs tv shows')

plt.title('Movies vs TV Shows')
plt.xlabel('Type')
plt.ylabel('Count')
plt.legend()
plt.show()
```

```python
df.fillna({'rating':'unknown'},inplace=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
rating_count = df['rating'].value_counts()
```

```python
plt.pie(rating_count.values, labels=rating_count.index, autopct='%1.1f%%', radius=2.5)

plt.title('Rating Chart')
plt.show()
```

```python
df.head()
```

```python
df.dropna(subset=['duration'], inplace=True)
```

```python
movies = df[df['type']=='Movie'].copy()
movies['duration_int'] = movies['duration'].str.replace(' min','').astype(int)
movies_duration = movies['duration_int']
```

```python
movies.head()
```

```python
movies_duration
```

```python
plt.hist(movies_duration, color='orangered', bins=10, edgecolor='gray')

plt.title('Movies Duration Graph')
plt.xlabel('Movie Duration')
plt.ylabel('Number of Movies')
plt.tight_layout()

plt.show()
```

```python
df.head()
```

```python
release_count = df['release_year'].value_counts().sort_index()
```

```python
plt.scatter(release_count.index, release_count.values, color='orangered', edgecolor='gray')

plt.title('Release Year Plot')
plt.xlabel('Release Years')
plt.ylabel('Number of Shows')
plt.tight_layout()
plt.show()
```

```python
release_type_by_year = df.groupby(['release_year','type']).size().unstack().fillna(0)
```

```python
plt.plot(release_type_by_year.index, release_type_by_year['Movie'], color='orange')
plt.plot(release_type_by_year.index, release_type_by_year['TV Show'], color='orangered')
plt.title('TV Show VS Movie per Year')
plt.xlabel('Years')
plt.ylabel('Number of Movies and TV Shows')
plt.show()
```
