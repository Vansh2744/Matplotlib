```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('netflix_users.csv')
df.head()
```

```python
countries = df['Country'].unique()
viewers = df['Country'].value_counts()
```

```python
plt.figure(figsize=(12, 6))
plt.bar(countries, viewers, color='orange')

plt.title('Netflix Viewers Data')
plt.xlabel('Country')
plt.ylabel('Viewers')

plt.savefig('netflix_viewers.png', dpi=300, bbox_inches='tight')
plt.show()
```
