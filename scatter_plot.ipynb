```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
study_time = np.array([8, 5, 6, 7, 3, 2, 5, 2, 3, 4, 10])
scores = np.array([90, 80, 85, 88, 56, 43, 56, 45, 50, 55, 98])
```

```python
plt.scatter(study_time, scores, color='orange', marker='o')
plt.title('Student Score')
plt.xlabel('Study Time')
plt.ylabel('Score')
plt.show()
```

```python
study_time1 = np.array([8, 5, 6, 7, 3, 2, 5, 2, 3, 4, 10])
scores1 = np.array([90, 80, 85, 88, 56, 43, 56, 45, 50, 55, 98])

study_time2 = np.array([8, 5, 6, 7, 3, 2, 5, 2, 3, 4, 10])
scores2 = np.array([80, 70, 80, 68, 65, 50, 46, 55, 30, 59, 90])
```

```python
plt.scatter(study_time1, scores1, color='red')
plt.scatter(study_time2, scores2, color='orange')
plt.title('Two classes comparison')
plt.xlabel('Study Time')
plt.ylabel('Score')
plt.grid()
plt.show()
```
