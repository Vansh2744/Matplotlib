```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('healthcare_dataset.csv')
df.head()
```

```python
blood_types = df['Admission Type'].unique()
patients = df['Admission Type'].value_counts()
```

```python
fig, ax = plt.subplots(1,2, figsize=(20, 10))

ax[0].plot(blood_types, patients, color='red')
ax[0].set_title('Line Plot', size=20)

ax[1].bar(blood_types, patients, width=0.5, color='orange')
ax[1].set_title('Bar Chart', size=20)

plt.suptitle('Comparison of Line Plot and Bar Plot', size=30)
plt.tight_layout()
plt.show()
```
