```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.array([10,20,30,40,50,60])
y = np.array([100,200,150,300,250,400])
```

```python
plt.plot(x,y)
plt.show()
```

```python
x = ['Vansh', 'Aman', 'Rahul', 'Kartik', 'Rohan']
y = [23, 45, 32, 67, 89]
```

```python
plt.plot(x,y)

plt.title('Score Plot')
plt.xlabel('Student Names')
plt.ylabel('Student Scores')
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 150, 200, 300, 250]
```

```python
plt.plot(x,y, color='red', linewidth=1, linestyle=':', marker='o', markersize=10, label='Sales')

plt.legend(loc='lower right')
plt.grid(color='cyan', linestyle=':', linewidth=1.5)
# plt.xlim(10, 100)
# plt.ylim(100, 1000)
plt.xticks([10,20,30,40,50],['1-10 weeks','11-20 weeks','21-30 weeks','31-40 weeks','41-50 weeks'])
plt.show()
```
